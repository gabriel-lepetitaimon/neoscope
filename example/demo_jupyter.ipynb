```python
from ipywidgets import Label

from neoscope.base import HelloButton

label = Label()


def on_click(value):
    print(f"Button clicked! Value: {value}")
    label.value = f"Button clicked! Value: {value}"


b = HelloButton()
b.on_click(on_click)
b.anywidget()
```

```python
b._dispatch_event("click", [5])  # Simulate a click event with value 42
```
